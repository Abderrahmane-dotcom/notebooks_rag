```python
# A Sample Class that will help to convert sentences into Token
class Tokenizer(object):
    def __init__(self):
        self.model = SentenceTransformer('all-MiniLM-L6-v2')

    def get_token(self, documents):
        sentences  = [documents]
        sentence_embeddings = self.model.encode(sentences)
        encod_np_array = np.array(sentence_embeddings)
        encod_list = encod_np_array.tolist()
        return encod_list[0]
```

```python
# Define Settings and mappings for the index
# Define timeout enough to avoid crashing

from elasticsearch import Elasticsearch

# Initialize the Elasticsearch client
# Make you have enough Timeout otherwise it will crash 
es = Elasticsearch("http://localhost:9200")

# Check connection
if not es.ping():
    print("Elasticsearch connection failed")
else:
    print("Connected to Elasticsearch")

# Define the index configuration with Elastiknn
index_config = {
    "settings": {
        "index": {
            "number_of_shards": 1,
            "number_of_replicas": 0
        }
    },
    "mappings": {
        "properties": {
            "vector": {
                "type": "elastiknn_dense_float_vector",
                "elastiknn": {
                    "dims": 384,
                    "model": "exact",
                    "similarity": "cosine",
                    "L": 40,
                    "k": 4
                }
            },
            "title": {
                "type": "text",
                "fields": {
                    "keyword": {
                        "type": "keyword",
                        "index": False
                    }
                }
            },
            "company": {
                "type": "keyword",
                "index": False
            },
            "location": {
                "type": "keyword",
                "index": False
            },
            "salary": {
                "type": "keyword",
                "index": False
            },
            "job_description": {
                "type": "keyword",
                "index": False
            }
        }
    }
}

# Create the index
index_name = "posting"
if es.indices.exists(index=index_name):
    es.indices.delete(index=index_name)  # Optional: delete the index if it exists
response = es.indices.create(index=index_name, body=index_config)

# Print response
print(response)
```
